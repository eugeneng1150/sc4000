```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_colwidth", 120)
```

```python
candidate_dirs = [
    Path("/kaggle/input/predict-ai-model-runtime"),  # Kaggle notebook
    Path("predict-ai-model-runtime"),                # local unzip folder
    Path("data/predict-ai-model-runtime"),           # common project data folder
    Path("data"),
    Path("."),
]

data_dir = next(
    (
        path
        for path in candidate_dirs
        if path.exists() and (list(path.glob("*.csv")) or list(path.rglob("*.npz")))
    ),
    None,
)

if data_dir is None:
    data_dir = Path("data/predict-ai-model-runtime")
    print("Competition data not found yet.")
    print(f"Put the unzipped Kaggle files here: {data_dir.resolve()}")
else:
    print(f"Using data directory: {data_dir.resolve()}")
```

```python
csv_files = sorted(data_dir.rglob("*.csv"))

csv_overview = pd.DataFrame(
    {
        "file": [str(path.relative_to(data_dir)) for path in csv_files],
        "size_mb": [path.stat().st_size / 1_000_000 for path in csv_files],
    }
)

csv_overview.head()
```

```python
sample_submission_path = data_dir / "sample_submission.csv"

if sample_submission_path.exists():
    sample_submission = pd.read_csv(sample_submission_path)
    display(sample_submission.head())
else:
    print("sample_submission.csv was not found in the selected data directory.")
```

```python
npz_files = sorted(data_dir.rglob("*.npz"))

npz_overview = pd.DataFrame(
    {
        "file": [str(path.relative_to(data_dir)) for path in npz_files],
        "size_mb": [path.stat().st_size / 1_000_000 for path in npz_files],
    }
)

npz_overview.head()
```

```python
if not npz_files:
    print("No .npz graph files found yet. Download and unzip the Kaggle competition data first.")
else:
    example_npz_path = npz_files[0]
    print(f"Inspecting: {example_npz_path.relative_to(data_dir)}")

    with np.load(example_npz_path, allow_pickle=True) as example_npz:
        npz_head = pd.DataFrame(
            {
                "key": example_npz.files,
                "shape": [example_npz[key].shape for key in example_npz.files],
                "dtype": [example_npz[key].dtype for key in example_npz.files],
            }
        )

    display(npz_head.head(20))
```
